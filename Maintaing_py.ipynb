 Normalization
 x' = x - x min / x max - x min

 suppose we have a dataset

 Blue  $70000 45
 Red   $60000 44
 Purple $52000 40
 Step-by-Step CalculationFeature 1: Numerical Values
 ($70,000, $60,000, $52,000)
 $x_{min}$: $52,000$$x_{max}$: $70,000$Range ($x_{max} - x_{min}$): $18,000$Blue: $\frac{70,000 - 52,000}{18,000} = \frac{18,000}{18,000} = \mathbf{1.0}$Red: $\frac{60,000 - 52,000}{18,000} = \frac{8,000}{18,000} \approx \mathbf{0.444}$Purple: $\frac{52,000 - 52,000}{18,000} = \frac{0}{18,000} = \mathbf{0.0}$Feature 2: Age/Count (45, 44, 40)$x_{min}$: $40$$x_{max}$: $45$Range ($x_{max} - x_{min}$): $5$Blue: $\frac{45 - 40}{5} = \frac{5}{5} = \mathbf{1.0}$Red: $\frac{44 - 40}{5} = \frac{4}{5} = \mathbf{0.8}$Purple: $\frac{40 - 40}{5} = \frac{0}{5} = \mathbf{0.0}$Normalized DatasetAfter applying the formula, your dataset is now on a uniform scale:ColorFeature 1 (Normalized)Feature 2 (Normalized)Blue1.0001.000Red0.4440.800Purple0.0000.000


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


df = pd.read_csv(r'/content/Data.csv')
print(df)
x = df.iloc[: ,: -1].values
print(x)
# it is taking all the range except last column
y = df.iloc[:, -1].values
print(y)
```

```python
import numpy as np
from sklearn.impute import SimpleImputer

# 1. Initialize the imputer (fixed the double '=' and added np import)
imputer = SimpleImputer(missing_values=np.nan, strategy='mean')

# 2. Fit the imputer to the specific columns (indices 1 and 2)
imputer.fit(x[:, 1:3])

# 3. Transform the data and assign it back to the array
x[:, 1:3] = imputer.transform(x[:, 1:3])

print(x)
print(y)


```

```python
import numpy as np
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# 1. Define the transformer
# We pass a list containing a tuple: (name, object, column_indices)
ct = ColumnTransformer(
    transformers=[('encoder', OneHotEncoder(), [0])],
    remainder='passthrough'
)

# 2. Fit and transform your data (and convert to a numpy array)
x = np.array(ct.fit_transform(x))
```
