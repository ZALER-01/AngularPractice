 Normalization
 x' = x - x min / x max - x min

 suppose we have a dataset

 Blue  $70000 45
 Red   $60000 44
 Purple $52000 40
 Step-by-Step CalculationFeature 1: Numerical Values
 ($70,000, $60,000, $52,000)
 $x_{min}$: $52,000$$x_{max}$: $70,000$Range ($x_{max} - x_{min}$): $18,000$Blue: $\frac{70,000 - 52,000}{18,000} = \frac{18,000}{18,000} = \mathbf{1.0}$Red: $\frac{60,000 - 52,000}{18,000} = \frac{8,000}{18,000} \approx \mathbf{0.444}$Purple: $\frac{52,000 - 52,000}{18,000} = \frac{0}{18,000} = \mathbf{0.0}$Feature 2: Age/Count (45, 44, 40)$x_{min}$: $40$$x_{max}$: $45$Range ($x_{max} - x_{min}$): $5$Blue: $\frac{45 - 40}{5} = \frac{5}{5} = \mathbf{1.0}$Red: $\frac{44 - 40}{5} = \frac{4}{5} = \mathbf{0.8}$Purple: $\frac{40 - 40}{5} = \frac{0}{5} = \mathbf{0.0}$Normalized DatasetAfter applying the formula, your dataset is now on a uniform scale:ColorFeature 1 (Normalized)Feature 2 (Normalized)Blue1.0001.000Red0.4440.800Purple0.0000.000
